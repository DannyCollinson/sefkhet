```python
import snaplog


for i, level in enumerate((-1, 0, 1, 9, 10, 11, 20, 30, 40, 50, 60, 61)):
    snaplog.record(
        "I <3 snaplog!",
        level=level,
        logger=(
            str(i),
            -10,
            {"handlers": ("stdout", {"level": -10}), "color": "level"},
        ),
    )
```
